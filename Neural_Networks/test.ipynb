```python
import torch 
import torch.nn as nn
import torch.nn.functional as F

emb_size = 10
head_size= 16
context_size = 8
```

```python
query = nn.Linear(in_features=emb_size, out_features=head_size, bias=False)
key = nn.Linear(in_features=emb_size, out_features=head_size, bias=False)
value = nn.Linear(in_features=emb_size, out_features=head_size, bias=False) 
tril = torch.tril(torch.ones((context_size, context_size)))
```

```python
input = torch.rand((4,context_size, emb_size))
B,T,C = input.shape
q = query(input)   # (B,T,H)
k = key(input)     # (B,T,H)
v = value(input)   # (B,T,H)
w = q @ k.transpose(1,2) * C **-0.5    # (B,T,H) @ (B,H,T) = (B,T,T)
w = w.masked_fill(tril == 0, float('-inf'))    # (B,T,T) without speaking with the future
w = F.softmax(w, dim=2) # (B,T,T) 
out = w @ v # (B,T,T) @ (B,T,H) = (B,T,H)
```
